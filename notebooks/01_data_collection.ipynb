# 01 – Data collection
Downloads the raw datasets into `data/raw/`.

**Download date:** 5 November 2026
The Fire Brigade data and the DWD "recent" weather file are updated daily, so a later download may differ slightly.

```python
import requests
from pathlib import Path
```

```python
raw_folder = Path("../data/raw")
raw_folder.mkdir(parents=True, exist_ok=True)

print(raw_folder.resolve())
print(raw_folder.exists())
```

```python
fire_brigade_base = "https://raw.githubusercontent.com/Berliner-Feuerwehr/BF-Open-Data/main/Datasets/Daily_Data/"
dwd_climate_base = "https://opendata.dwd.de/climate_environment/CDC/observations_germany/climate/daily/kl/"
dwd_warnings_base = "https://opendata.dwd.de/climate_environment/health/historical_alerts/heat_warnings/"

urls = [
    fire_brigade_base + "BFw_mission_data_daily.csv",
    fire_brigade_base + "BFw_daily_mission_and_regional_data_descriptions.csv",
    dwd_climate_base + "historical/tageswerte_KL_00433_19480101_20251231_hist.zip",
    dwd_climate_base + "recent/tageswerte_KL_00433_akt.zip",
    dwd_warnings_base + "Description_historical_heat_alerts.pdf",
]

for year in range(2018, 2027):
    urls.append(f"{dwd_warnings_base}heat_alerts_{year}.csv")

print(len(urls), "files to download")
```

```python
for url in urls:
    file_name = url.split("/")[-1]
    response = requests.get(url, timeout=60)
    response.raise_for_status()

    if not response.content:
        raise ValueError(f"Empty download: {file_name}")

    (raw_folder / file_name).write_bytes(response.content)
    print(f"OK  {file_name}  ({len(response.content):,} bytes)")
```

```python
for file in sorted(raw_folder.iterdir()):
    size_kb = file.stat().st_size / 1024
    print(f"{file.name:60} {size_kb:8.1f} KB")
```
